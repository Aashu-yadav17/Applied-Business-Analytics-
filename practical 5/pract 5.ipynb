5. Probability and Probability Distributions:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
# For reproducible results
np.random.seed(42)
```

5A

```python


# 1. Simulate 1,000 dice rolls
dice_rolls = np.random.randint(1, 7, 1000)
print("\n1. First 20 dice roll outcomes:")
print(dice_rolls[:20])
# Frequency of each outcome
frequency = Counter(dice_rolls)
print("\nFrequency of each number:")
for number in range(1, 7):
    print(f"Number {number}: {frequency[number]}")
```

```python

# 2. Experimental probability
experimental_probability = {}

print("\n2. Experimental Probability:")
for number in range(1, 7):
    probability = frequency[number] / 1000
    experimental_probability[number] = probability
    print(f"P({number}) = {probability:.4f}")
```

```python

# 3. Theoretical probability
theoretical_probability = {number: 1/6 for number in range(1, 7)}
print("\n3. Theoretical Probability:")
for number in range(1, 7):
    print(f"P({number}) = {theoretical_probability[number]:.4f}")
```

```python
# 4. Probability of getting 1, 2, 3, 4, 5, 6
print("\n4. Probability of getting each number:")
for number in range(1, 7):
    print(f"P({number}) = 1/6 = {1/6:.4f}")
```

```python
# 5. Probability of an even number
even_count = np.sum(dice_rolls % 2 == 0)
even_probability = even_count / 1000

print("\n5. Probability of getting an even number:")
print(f"Experimental: {even_probability:.4f}")
print(f"Theoretical: {3/6:.4f}")
```

```python
# 6. Probability of an odd number
odd_count = np.sum(dice_rolls % 2 != 0)
odd_probability = odd_count / 1000
print("\n6. Probability of getting an odd number:")
print(f"Experimental: {odd_probability:.4f}")
print(f"Theoretical: {3/6:.4f}")
```

```python
# 7. Probability of number greater than 4
greater_4_count = np.sum(dice_rolls > 4)
greater_4_probability = greater_4_count / 1000
print("\n7. Probability of getting a number greater than 4:")
print(f"Experimental: {greater_4_probability:.4f}")
print(f"Theoretical: {2/6:.4f}")
```

```python
# 8. Probability of number less than 4
less_4_count = np.sum(dice_rolls < 4)
less_4_probability = less_4_count / 1000
print("\n8. Probability of getting a number less than 4:")
print(f"Experimental: {less_4_probability:.4f}")
print(f"Theoretical: {2/6:.4f}")
```

```python
# 9. Probability of number greater than or equal to 4
greater_equal_4_count = np.sum(dice_rolls >= 4)
greater_equal_4_probability = greater_equal_4_count / 1000
print("\n9. Probability of getting a number greater than or equal to 4:")
print(f"Experimental: {greater_equal_4_probability:.4f}")
print(f"Theoretical: {3/6:.4f}")
```

```python
# 10. BAR CHART - FREQUENCY
numbers = np.arange(1, 7)
frequencies = [frequency[i] for i in numbers]
plt.figure(figsize=(8, 5))
plt.bar(numbers, frequencies, color="skyblue", edgecolor="black")
plt.xlabel("Dice Outcome")
plt.ylabel("Frequency")
plt.title("Frequency of Dice Outcomes (1,000 Rolls)")
plt.xticks(numbers)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python

# 11. BAR CHART - EXPERIMENTAL PROBABILITY
probabilities = [experimental_probability[i] for i in numbers]
plt.figure(figsize=(8, 5))
plt.bar(numbers, probabilities, color="lightgreen", edgecolor="black")
plt.xlabel("Dice Outcome")
plt.ylabel("Experimental Probability")
plt.title("Experimental Probability of Dice Outcomes")
plt.xticks(numbers)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# 12. PIE CHART - DICE OUTCOMES
plt.figure(figsize=(7, 7))
plt.pie(
    frequencies,
    labels=numbers,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Percentage Distribution of Dice Outcomes")
plt.show()
```

```python
# 13. LINE CHART - EXPERIMENTAL VS THEORETICAL
theoretical_probabilities = [1/6] * 6
plt.figure(figsize=(9, 5))
plt.plot(
    numbers,
    probabilities,
    marker="o",
    linewidth=2,
    label="Experimental Probability"
)
plt.plot(
    numbers,
    theoretical_probabilities,
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Theoretical Probability"
)
plt.xlabel("Dice Outcome")
plt.ylabel("Probability")
plt.title("Experimental vs Theoretical Probability")
plt.xticks(numbers)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# 14. BAR CHART - EVEN VS ODD
even_odd_labels = ["Even", "Odd"]
even_odd_frequency = [even_count, odd_count]
plt.figure(figsize=(7, 5))
plt.bar(
    even_odd_labels,
    even_odd_frequency,
    color=["orange", "purple"],
    edgecolor="black"
)
plt.xlabel("Outcome Type")
plt.ylabel("Frequency")
plt.title("Frequency of Even vs Odd Outcomes")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python

# 15. EXPERIMENTAL PROBABILITY COMPARISON
categories = [
    "Even",
    "Odd",
    "> 4",
    "< 4"
]
category_probabilities = [
    even_probability,
    odd_probability,
    greater_4_probability,
    less_4_probability
]
plt.figure(figsize=(9, 5))
bars = plt.bar(
    categories,
    category_probabilities,
    color=["orange", "purple", "green", "blue"],
    edgecolor="black"
)
plt.xlabel("Category")
plt.ylabel("Experimental Probability")
plt.title("Experimental Probability of Different Dice Events")
plt.ylim(0, 1)
for bar, value in zip(bars, category_probabilities):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        value + 0.02,
        f"{value:.3f}",
        ha="center"
    )
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python

# NORMAL DISTRIBUTION
# 1. Generate 1,000 random values
normal_data = np.random.normal(
    loc=50,
    scale=10,
    size=1000
)
print("\n1. Normal Distribution")
print("Mean = 50")
print("Standard Deviation = 10")
print("Number of values = 1000")

# Calculate mode
# For continuous random data, exact values almost never repeat.
# We therefore report the rounded mode.
rounded_normal = np.round(normal_data, 1)
mode_normal = Counter(rounded_normal).most_common(1)[0][0]
```

```python

# 2 & 3. Statistics
normal_mean = np.mean(normal_data)
normal_median = np.median(normal_data)
normal_variance = np.var(normal_data)
normal_std = np.std(normal_data)
normal_min = np.min(normal_data)
normal_max = np.max(normal_data)

print("\n2 & 3. Normal Distribution Statistics")
print(f"Mean       : {normal_mean:.4f}")
print(f"Median     : {normal_median:.4f}")
print(f"Mode       : approximately {mode_normal}")
print(f"Variance   : {normal_variance:.4f}")
print(f"Std Dev    : {normal_std:.4f}")
print(f"Minimum    : {normal_min:.4f}")
print(f"Maximum    : {normal_max:.4f}")
```

```python
# 4. Histogram
plt.figure(figsize=(9, 5))
plt.hist(
    normal_data,
    bins=30,
    color="steelblue",
    edgecolor="black",
    alpha=0.75
)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normal Distribution")
plt.grid(axis="y", alpha=0.3)
plt.show()

print("\n4. Analysis of Normal Distribution:")
print("The histogram should have an approximately bell-shaped and symmetric form.")
print("Most values are concentrated around the mean of 50.")
print("Fewer values occur as we move farther away from the mean.")
```

```python
# 5. Box plot
plt.figure(figsize=(8, 4))
plt.boxplot(
    normal_data,
    vert=False,
    patch_artist=True,
    boxprops=dict(facecolor="lightblue")
)
plt.xlabel("Value")
plt.title("Box Plot of Normal Distribution")
plt.grid(axis="x", alpha=0.3)
plt.show()
```

```python
# 6. Percentage between 40 and 60
between_40_60 = np.mean(
    (normal_data >= 40) & (normal_data <= 60)
) * 100

print("\n6. Percentage of values where 40 <= X <= 60:")
print(f"{between_40_60:.2f}%")
```

```python
# 7. Percentage between 30 and 70
between_30_70 = np.mean(
    (normal_data >= 30) & (normal_data <= 70)
) * 100
print("\n7. Percentage of values where 30 <= X <= 70:")
print(f"{between_30_70:.2f}%")
```

```python
# 8. Generate 1,000 values from Uniform(0,100)
uniform_data = np.random.uniform(
    low=0,
    high=100,
    size=1000
)
print("\n8. Uniform Distribution")
print("Generated 1,000 random numbers between 0 and 100.")

```

```python

# 9. Histogram
plt.figure(figsize=(9, 5))
plt.hist(
    uniform_data,
    bins=30,
    color="mediumseagreen",
    edgecolor="black"
)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Uniform Distribution")
plt.grid(axis="y", alpha=0.3)
plt.show()

print("\n9. Uniform Distribution Analysis:")
print("The histogram should be approximately flat.")
print("Values throughout the interval 0 to 100 have approximately equal likelihood.")
```

```python
# 10. Box plot
plt.figure(figsize=(8, 4))
plt.boxplot(
    uniform_data,
    vert=False,
    patch_artist=True,
    boxprops=dict(facecolor="lightgreen")
)
plt.xlabel("Value")
plt.title("Box Plot of Uniform Distribution")
plt.grid(axis="x", alpha=0.3)
plt.show()
```

```python
# 11. Generate 1,000 values with scale = 2
exponential_data = np.random.exponential(
    scale=2,
    size=1000
)

print("\n11. Exponential Distribution")
print("Generated 1,000 values with scale = 2.")
```

```python
# 12. Histogram
plt.figure(figsize=(9, 5))
plt.hist(
    exponential_data,
    bins=30,
    color="salmon",
    edgecolor="black"
)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Exponential Distribution")
plt.grid(axis="y", alpha=0.3)
plt.show()
print("\n12. Exponential Distribution Analysis:")
print("The distribution is strongly right-skewed.")
print("Most values are close to zero, while a small number of values are much larger.")
```

```python
# 13. Box plot
plt.figure(figsize=(8, 4))
plt.boxplot(
    exponential_data,
    vert=False,
    patch_artist=True,
    boxprops=dict(facecolor="lightcoral")
)
plt.xlabel("Value")
plt.title("Box Plot of Exponential Distribution")
plt.grid(axis="x", alpha=0.3)
plt.show()

Q1 = np.percentile(exponential_data, 25)
Q3 = np.percentile(exponential_data, 75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

extreme_values = exponential_data[
    (exponential_data < lower_bound) |
    (exponential_data > upper_bound)
]

print("\n13. Possible Extreme Values:")
print(f"Q1 = {Q1:.4f}")
print(f"Q3 = {Q3:.4f}")
print(f"IQR = {IQR:.4f}")
print(f"Upper outlier boundary = {upper_bound:.4f}")
print(f"Number of possible extreme values = {len(extreme_values)}")
```

```python
# 14. Percentage greater than 2
greater_2 = np.mean(exponential_data > 2) * 100

print("\n14. Percentage of values greater than 2:")
print(f"{greater_2:.2f}%")
```

```python
# 15. Percentage greater than 5
greater_5 = np.mean(exponential_data > 5) * 100
print("\n15. Percentage of values greater than 5:")
print(f"{greater_5:.2f}%")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulate 10,000 coin tosses
coin_tosses_10000 = np.random.choice(
    ["Heads", "Tails"],
    size=10000
)

# Count Heads and Tails
heads_10000 = np.sum(coin_tosses_10000 == "Heads")
tails_10000 = np.sum(coin_tosses_10000 == "Tails")

# Experimental probabilities
experimental_heads_10000 = heads_10000 / 10000
experimental_tails_10000 = tails_10000 / 10000

# Theoretical probabilities
theoretical_heads = 0.5
theoretical_tails = 0.5

# Difference between experimental and theoretical probabilities
difference_heads = experimental_heads_10000 - theoretical_heads
difference_tails = experimental_tails_10000 - theoretical_tails

# Print complete analysis
print("=" * 50)
print("COMPLETE ANALYSIS - 10,000 COIN TOSSES")
print("=" * 50)

print("Total tosses:", 10000)
print("Number of Heads:", heads_10000)
print("Number of Tails:", tails_10000)

print("\nExperimental Probabilities:")
print("Heads:", round(experimental_heads_10000, 4))
print("Tails:", round(experimental_tails_10000, 4))

print("\nTheoretical Probabilities:")
print("Heads:", theoretical_heads)
print("Tails:", theoretical_tails)

print("\nDifference:")
print("Heads:", round(difference_heads, 4))
print("Tails:", round(difference_tails, 4))

# Bar chart
plt.figure(figsize=(7, 5))

plt.bar(
    ["Heads", "Tails"],
    [heads_10000, tails_10000],
    color=["gold", "lightgray"],
    edgecolor="black"
)

plt.xlabel("Outcome")
plt.ylabel("Frequency")
plt.title("Heads vs Tails - 10,000 Coin Tosses")

plt.show()

# Pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    [heads_10000, tails_10000],
    labels=["Heads", "Tails"],
    autopct="%1.1f%%",
    colors=["gold", "lightgray"],
    startangle=90
)

plt.title("Percentage Distribution - 10,000 Coin Tosses")

plt.show()
```
